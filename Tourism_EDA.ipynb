- 2024·2025년은 4월 이후 6월까지 증가하지만, 2026년은 5월에 증가한 뒤 6월에 다시 감소한다.
- 방문자수에서 발견한 ‘6월 재감소’가 순 방문자수에서는 세 해 모두 반복되지 않는다.

```python
봄여름_비교 = df1[df1['월'].isin([3, 4, 5, 6])]

display(봄여름_비교[['연도', '월', '방문자수', '순 방문자수']])
```

```python
분석자료 = df1[df1['연도'].isin([2024, 2025])]
```

```python
월별평균 = 분석자료.groupby('월')[
    ['방문자수', '순 방문자수']
].mean()

월별평균 = 월별평균.reset_index()

display(월별평균.round(0))
```

```python
print(월별평균['방문자수'].max())
print(월별평균['방문자수'].min())

print(월별평균['순 방문자수'].max())
print(월별평균['순 방문자수'].min())
```

## 해석
모든 월에 동일하게 2년치 자료를 반영하기 위해, 월별 평균 비교에는 2024·2025년만 사용하였다.


- 가장 높은 방문자수를 기록한 달은 8월, 낮은 수를 기록한 달은 4월
- 가장 높은 순 방문자수를 기록한 달은 2월, 낮은 수를 기록한 달은 4월

```python
최저_행번호 = df1.groupby('연도')['방문자수'].idxmin()

연도별_최저 = df1.loc[
    최저_행번호,
    ['연도', '월', '방문자수']
]

display(연도별_최저)
```

```python
최저_행번호 = df1.groupby('연도')['순 방문자수'].idxmin()

연도별_최저 = df1.loc[
    최저_행번호,
    ['연도', '월', '순 방문자수']
]

display(연도별_최저)
```

```python
계절자료 = df1[df1['연도'].isin([2024, 2025])].copy()

월별_계절 = {
    3: '봄', 4: '봄', 5: '봄',
    6: '여름', 7: '여름', 8: '여름',
    9: '가을', 10: '가을', 11: '가을',
    12: '겨울', 1: '겨울', 2: '겨울'
}

계절자료['계절'] = 계절자료['월'].map(월별_계절)
```

```python
계절별평균 = 계절자료.groupby(['연도', '계절'])[
    ['방문자수', '순 방문자수']
].mean()

display(계절별평균.round(0))
```

계절별 차이는 관찰되었지만, 두 해의 자료만으로 일반적인 계절성을 단정할 수 없다.

# 2. 유료관광지별 입장객 집중도 분석
- df3
- 수집된 9곳의 유료관광지 중 입장객이 어디에 집중되는가?
- 2026년에는 입장객수 결측치가 있으므로, 이번 비교는 입장객수의 결측값(NaN)이 없는 2024~2025년 자료를 사용한다.

```python
관광지_분석자료 = df3[
    (df3['기준연월'] >= 202401) &
    (df3['기준연월'] <= 202512)
].copy()

관광지별_입장객수 = 관광지_분석자료.groupby(
    ['관광지명', '기초지자체명'],
    as_index=False
)['입장객수(명)'].sum()

관광지별_입장객수 = 관광지별_입장객수.sort_values(
    '입장객수(명)',
    ascending=False
)

display(관광지별_입장객수)
```

```python
import plotly.express as px

fig = px.bar(
    data_frame=관광지별_입장객수,
    x='관광지명',
    y='입장객수(명)',
    color='기초지자체명',
    title='부산 9개 유료관광지별 입장객수 합계 (2024~2025년)',
    labels={
        '관광지명': '관광지',
        '입장객수(명)': '입장객수 합계(명)',
        '기초지자체명': '구·군'
    }
)

fig.update_xaxes(categoryorder='total descending')
fig.update_yaxes(tickformat=',.0f', hoverformat=',.0f')

fig.show()
```

```python
# 9개 관광지의 입장객수 합계
전체_입장객수 = 관광지별_입장객수['입장객수(명)'].sum()

상위4곳 = 관광지별_입장객수.head(4)

상위4곳_입장객수 = 상위4곳['입장객수(명)'].sum()

상위4곳_비중 = (상위4곳_입장객수 / 전체_입장객수) * 100

display(상위4곳)

print(f'9개 관광지 전체 입장객수: {전체_입장객수:,.0f}명')
print(f'상위 4곳 입장객수: {상위4곳_입장객수:,.0f}명')
print(f'상위 4곳의 비중: {상위4곳_비중:.1f}%')
```

## 해석

- 2024ㆍ2025년 9곳의 유료관광지별 입장객수에서 '국립부산과학관'의 입장객수가 가장 많았고, 구포국수체험관의 입장객수가 가장 낮은 기록을 보였다.

- 국립부산과학관, 부산아쿠아리움, 한국마사회 렛츠런파크 부산경남, 태종대 관광지의 입장객수 합계는 수집된 9개 관광지 전체의 약 79.4%를 차지했다.

- 하지만, 이 수치는 관광지 입장 건수의 누적 합계이므로, 중복을 제거한 관광객 수나 부산 전체 관광객의 지역별 분포로 해석할 수 없다.

## 해석

- 2024년 1월~2026년 10월의 월별 자료에서 숙박자 비율과의 상관계수는 **순 방문자수 0.175, 방문자수 0.298**로 나타났다.

- **두 지표 모두 약한 양의 선형 관계**를 보여, 방문 규모가 큰 달에 숙박자 비율도 뚜렷하게 높아진다고 보기는 어려웠다. **방문 규모만으로 숙박자 비율의 차이를 설명하는 데에는 한계가 있다.**

- 산점도에서도 순 방문자수가 증가할수록 숙박자 비율이 일관되게 상승하거나 하락하는 모습은 뚜렷하지 않았다. 비슷한 방문 규모에서도 숙박자 비율에 차이가 나타났으며, 이는 약한 양의 상관계수(0.175)와 부합한다.



## 해석의 한계

이 결과는 월별 방문 규모와 숙박자 비율의 관계를 살펴본 것으로, 개별 방문객의 숙박 전환 과정이나 실제 체류시간을 확인한 것은 아니다. 또한 방문객 증가가 숙박자 비율 상승을 유발한다는 인과관계를 의미하지 않는다.

# 4.관광소비 및 방문 규모 대비 소비액 분석

- 방문 규모를 고려해도 소비가 높은 달은 언제인가?

```python
월별_관광소비 = df2[
    df2['업종대분류명'] == '전체'
][['기준연월', '소비액(천원)']].copy()

월별_관광소비 = 월별_관광소비.sort_values('기준연월')

display(월별_관광소비.head())

print('월별 관광소비 행 수:', len(월별_관광소비))
```

```python
소비_분석자료 = 소비_분석자료.sort_values('기준연월_변환')

소비_분석자료['관광소비액(억원)'] = (
    소비_분석자료['소비액(천원)'] / 100000
)

fig = px.line(
    data_frame=소비_분석자료,
    x='기준연월_변환',
    y='관광소비액(억원)',
    title='부산 월별 관광소비액 변화',
    markers=True,
    labels={
        '기준연월_변환': '기준연월',
        '관광소비액(억원)': '관광소비액(억원)'
    }
)

fig.update_xaxes(
    dtick='M3',
    tickformat='%Y년 %m월',
    hoverformat='%Y년 %m월'
)

fig.update_yaxes(
    tickformat=',.1f',
    hoverformat=',.1f',
    rangemode='tozero'
)

fig.show()
```

```python
소비_분석자료['관광소비액(원)'] = (
    소비_분석자료['소비액(천원)'] * 1000
)

소비_분석자료['순 방문자수 대비 소비액(원)'] = (
    소비_분석자료['관광소비액(원)']
    / 소비_분석자료['순 방문자수']
)

소비지표_순위 = 소비_분석자료.sort_values(
    '순 방문자수 대비 소비액(원)',
    ascending=False
)

display(
    소비지표_순위[
        ['기준연월', '순 방문자수', '관광소비액(억원)',
         '순 방문자수 대비 소비액(원)']
    ].head(5).round(1)
)
```

 ## 해석

순 방문자수와 관광소비액의 상관계수는 0.457로, 방문 규모가 큰 달에 관광소비액도 높은 경향이 나타났다. 다만 방문 규모만으로 소비액의 차이를 모두 설명할 수는 없다.

관광소비액이 가장 높은 달은 2026년 5월이었지만, 순 방문자수 대비 소비액이 가장 높은 달은 2025년 1월이었다. 총소비액이 높은 달과 방문 규모 대비 소비 지표가 높은 달은 서로 다를 수 있음을 확인했다.

## 해석의 한계

순 방문자수 대비 소비액은 서로 다른 자료에 기반한 월별 비교용 참고 지표이며, 관광객 한 사람이 실제 지출한 평균 금액을 의미하지 않는다.

이 지표는 소비액이 커지거나 방문자수가 작아져도 높아질 수 있으므로, 값이 높다는 이유만으로 소비 활성화 성과가 우수하다고 단정할 수 없다. 또한 상관관계는 방문객 증가가 소비 증가를 유발했다는 인과관계를 입증하지 않는다.
